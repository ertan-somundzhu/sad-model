```python
import os

from pydub import AudioSegment
AudioSegment.converter = "/usr/bin/ffmpeg"

import numpy as np
from numpy.lib.stride_tricks import sliding_window_view

import torch

import matplotlib.pyplot as plt

from logMelSpectrogram import logMelSpectrogram

import json
from tqdm import tqdm
import random

random.seed(42)
```

```python
from huggingface_hub import snapshot_download
snapshot_download('nccratliri/vad-human-ava-speech', local_dir = "data/human-ava-speech", repo_type="dataset" )
```

```python
sr = 16000
stride_s = 10
frame_rate_s = 25

path = "/content/data/human-ava-speech/train/"
files = sorted(os.listdir(path))

filenames = []
log_mel_spec_data = []
onset_data = []
offset_data = []

n_files = len(files) // 2
range_n_files = random.sample(range(n_files), int(n_files * 0.25))

log_mel_spec = logMelSpectrogram()

stride = sr * stride_s // 1000
frame_rate = sr * frame_rate_s // 1000
```

```python
for i in tqdm(range_n_files):


    index = i * 2
    json_filename = os.path.join(path, files[index])
    audio_filename = os.path.join(path, files[index + 1])


    audio = AudioSegment.from_file(audio_filename)
    audio = audio.set_channels(1).set_frame_rate(sr)
    samples = np.array(audio.get_array_of_samples(), dtype=np.float32)


    log_mel = log_mel_spec.transform(samples=samples, sr=sr).cpu().numpy()
    n_frames = log_mel.shape[1]

    with open(json_filename) as f:
        d = json.load(f)


    onset_array = np.zeros(n_frames, dtype=np.int8)
    offset_array = np.zeros(n_frames, dtype=np.int8)

    onsets = (sr * np.array(d["onset"]) // 1000).astype(int)
    offsets = (sr * np.array(d["offset"]) // 1000).astype(int)

    onsets = onsets[(onsets >= 0) & (onsets < n_frames)]
    offsets = offsets[(offsets >= 0) & (offsets < n_frames)]

    onset_array[onsets] = 1
    offset_array[offsets] = 1


    chunks_mel = sliding_window_view(log_mel, window_shape=frame_rate, axis=1)[:, ::stride, :]
    chunks_mel = np.transpose(chunks_mel, (1, 0, 2))
    chunks_onset = sliding_window_view(onset_array, window_shape=frame_rate, axis=0)[::stride, :]
    chunks_offset = sliding_window_view(offset_array, window_shape=frame_rate, axis=0)[::stride, :]

    # to avoid data leakage from train to val samples
    files_chunks = np.full((chunks_offset.shape[0], 1), files[index + 1].split(".")[0])

    log_mel_spec_data.append(chunks_mel)
    onset_data.append(chunks_onset)
    offset_data.append(chunks_offset)
    filenames.append(files_chunks)


log_mel_spec_data = np.vstack(log_mel_spec_data)
onset_data = np.vstack(onset_data)
offset_data = np.vstack(offset_data)
filenames = np.vstack(filenames)
```

```python
np.savez(
    "dataset.npz",
    mel=log_mel_spec_data,
    onset=onset_data,
    offset=offset_data,
    filenames=filenames
)
```
